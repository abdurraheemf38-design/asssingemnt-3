# atm menu system 

```python
balance = 50000

while True:
    print(''' 1. check balance 
2. deposit money
3. withdraw money
4. exit ''')
    user = int(input(" select options from menu"))
    if user == 1 :
        print (f" the balance is {balance}")

    elif user == 4:
        print("menu closed")
        break

    elif user == 2 :
        deposit = int(input("enter the amount you want to deposit "))
        balance += deposit 

    elif user == 3 :
        withdraw =  int(input("enter the amount you want to withdraw "))
        if withdraw > balance :
            print(" you can't withdraw your amount is invalid ")
            
        else :
             balance -= withdraw
            
            
            
       
          
```

# guess the number 


```python
secret_num = 25
attempt = 0

while True:
    attempt +=1
    user = int(input("guess the number :"))
    if    user > secret_num :
        print(f"{user} guess is too high")

    elif user < secret_num :
        print(f"{user} guess is too low") 

    elif user == secret_num:
        print(f"{user} guess is correct")
        break
print(f"you made attempts: {attempt}")

        
    
```

# student performance  management system 



```python
total = 0
passed = 0
failed = 0
percent=[]

no_of_student = int(input(" enter no of students"))

for i in range (1,no_of_student+1):
    total+=1
    
    name = input("tell your name : ")
    python = int(input(" enter marks of python : "))
    maths  = int(input(" enter marks of maths :"))
    english = int(input(" enter marks of english :"))
    total_marks = python + maths + english 
    percentage = total_marks/300 *100
    percent.append(percentage)
    avg = sum(percent)/ no_of_student
    high = max(percent)
    low = min (percent)

    if python >= 40 and maths >= 40 and english >= 40 :
        if percentage >= 50:
            passed+=1
            print(" pass")
            if  percentage>= 90 :
                print("A+")
            elif percentage>= 80  :
                print("A")
            elif percentage>= 70 :
                print("B")
            elif  percentage>= 60 :
                print("C")
            elif percentage >= 50  :
                print("D")
        else:
            print("fail : must have 50 or above percentage ")
            failed+=1
    else :
        print(" fail: must have  40 or  above marks in all 3 subjects ")
        failed+=1




print(f'''
================= CLASS REPORT ====================

total students : {total}
passed students :{passed}
failed students :{failed}
highest percenatge :{high}
lowest percentage :{low}
average percentage {avg} ''')





```
